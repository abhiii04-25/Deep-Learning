# Your First Neural Network with Keras
Train a small model on MNIST handwritten digits. Run cells in order and explain each step in your own words.

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
keras.utils.set_random_seed(42)
```

## 1. Load and prepare data
MNIST images are 28×28 grayscale images. Scale pixels from 0–255 to 0–1.

```python
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()
x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0
print(x_train.shape, y_train.shape, x_test.shape)
```

## 2. Build the model
Flatten turns each image into a vector. The output layer has 10 units for digits 0–9.

```python
model = keras.Sequential([
    keras.Input(shape=(28, 28)),
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.2),
    layers.Dense(10, activation='softmax'),
])
model.summary()
```

```python
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
history = model.fit(x_train, y_train, epochs=5, batch_size=128, validation_split=0.1)
```

## 3. Evaluate and save
Keep test data separate from tuning decisions.

```python
test_loss, test_accuracy = model.evaluate(x_test, y_test, verbose=0)
print(f'Test loss: {test_loss:.4f}; test accuracy: {test_accuracy:.4f}')
model.save('mnist_model.keras')
```
